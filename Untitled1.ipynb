```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#define the antenna system
wavelegnth=1.0
spacing=wavelegnth/2
#waves arriving from -90 to 90 degrees
angles=np.linspace(-90,90,721)
theta=np.radians(angles)
```

```python
#calculating phase difference at each direction
phase_difference=(2*np.pi*spacing/wavelegnth)*np.sin(theta)
```

```python
#combine the signals from both antennas
combined=np.abs((1+np.exp(1j*phase_difference))/2)
```

```python
#plot the result
plt.figure(figsize=(9,4.5))
plt.plot(angles,combined)
plt.xlabel("Incoming wave angle(degrees)")
plt.ylabel("Normalised combined amplitude")
plt.title("Two_Element Receiving Array")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```
